Cree un bucle que sume los números del 100 al 200

```python
# Inicializar una variable para almacenar la suma
suma_total = 0

# Utilizar un bucle for para iterar sobre los números del 100 al 200
for numero in range(100, 201):
    # Sumar cada número al total
    suma_total += numero

# Imprimir el resultado
print("La suma de los números del 100 al 200 es:", suma_total)
```

Mostrar series de Fibonacci hasta 10 términos 

```python
# Definir los dos primeros términos de la serie de Fibonacci
fibonacci = [0, 1]

# Utilizar un bucle for para calcular y agregar los siguientes términos
for i in range(2, 10):  # Comenzamos desde el tercer término (índice 2)
    siguiente = fibonacci[i-1] + fibonacci[i-2]
    fibonacci.append(siguiente)

# Imprimir la serie de Fibonacci
print("Los primeros 10 términos de la serie de Fibonacci son:")
for termino in fibonacci:
    print(termino)
```

Escriba un programa en Python para imprimir todos los números primos entre 0 y 100 e imprima cuántos números primos hay.

```python
# Inicializar una lista para almacenar los números primos encontrados
numeros_primos = []

# Iterar sobre los números del rango de 0 a 100
for num in range(2, 101):
    es_primo = True
    
    # Verificar si el número es primo
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            es_primo = False
            break
    
    # Si el número es primo, agregarlo a la lista
    if es_primo:
        numeros_primos.append(num)

# Imprimir los números primos encontrados
print("Los números primos entre 0 y 100 son:")
for primo in numeros_primos:
    print(primo, end=" ")

# Imprimir el total de números primos encontrados
print("\nTotal de números primos:", len(numeros_primos))
```
