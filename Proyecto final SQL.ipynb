```python
import pandas as pd
import numpy as np

df = pd.read_csv('events.csv')
df
```

```python
df.dropna()
```

```python
df = df.replace({np.nan:''})
```

```python
#conexion con la base de datos
conexion = mysql.connector.connect(
    host = "localhost",
    user = "root",
    password = "Mark58913106**!!++",
    database = "proyecto_final_sql")


#creación de una tabla
try:
    for i in range(0, df.shape[0]):
        event_time = df.loc[i,'event_time']
        event_type = df.loc[i,'event_type']
        product_id = df.loc[i,'product_id']
        category_id = df.loc[i,'category_id']
        category_code = df.loc[i,'category_code']
        brand = df.loc[i,'brand']
        price = df.loc[i,'price']
        user_id = df.loc[i,'user_id']
        user_session = df.loc[i,'user_session']
        query = "INSERT INTO events(event_time, event_type, product_id, category_id, category_code, brand, price, user_id, user_session) values(" + event_time + "," + event_type + "," + str(product_id) + "," + str(category_id) + "," + category_code + "," + brand + "," + str(price) + "," + str(user_id) + "," + user_session + ");"
        cursor = conexion.cursor()
        cursor.execute(query)
        conexion.commit()
        print("Se ingresó correctamente el registro:", i)
        cursor.close()
except:
    print("Ocurrió un error al intentar ingresar el registro. Inténtelo de nuevo.")
```
